This notebook demonstrates how to collect course information from the GeeksforGeeks Courses webpage using `Requests` and `BeautifulSoup`, extract relevant HTML elements, organize the data in a Pandas DataFrame, and export the result to a CSV file.

## 2. Import Required Libraries

```python
# NumPy and Pandas are used for data handling
# Requests is used to fetch the webpage
# BeautifulSoup is used to parse and extract HTML content
import numpy as np
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

## 3. Send a Request to the Website

First, the webpage is requested and its HTTP response status code is checked. A status code of `200` indicates a successful response.

```python
# URL of the GeeksforGeeks Courses page
link = "https://www.geeksforgeeks.org/courses"

# Send a GET request to the webpage
data = requests.get(link)

# Display the HTTP response status code
print(data.status_code)
```

## 4. Use Request Headers

A User-Agent header is added so that the request resembles a request from a normal web browser.

```python
# Define a browser-like User-Agent
headers = {"User-Agent": "Mozilla/5.0"}

# Send the request with the headers
data = requests.get(link, headers=headers)
print(data.status_code)
```

## 5. Parse the HTML using BeautifulSoup

`lxml` is used as the HTML parser. BeautifulSoup converts the raw HTML response into a searchable structure.

```python
# Parse the webpage HTML using the lxml parser
soup = BeautifulSoup(data.text, "lxml")

# Display the formatted HTML structure
print(soup.prettify())
```

## 6. Extract the Main Page Heading

```python
# Find all H1 elements on the page
soup.find_all("h1")
```

```python
# Display the text of the first H1 element
soup.find_all("h1")[0].text
```

## 7. Extract Course Names

Course headings are located using the `h4` HTML tag.

```python
# Inspect all H4 elements containing course headings
for i in soup.find_all("h4"):
    print(i)
```

```python
# Display the text contained in each course heading
for i in soup.find_all("h4"):
    print(i.text)
```

```python
# Create a list to store course names
course_heading = []

# Extract and store each course heading
for i in soup.find_all("h4"):
    course_heading.append(i.text)

course_heading
```

## 8. Extract Course Ratings

Course rating information is extracted from `span` elements with the `urw-din` class.

```python
# Inspect span elements available on the webpage
soup.find_all("span")
```

```python
# Display the text from span elements with the required class
for i in soup.find_all("span", class_="urw-din"):
    print(i.text)
```

```python
# Create a list to store course ratings
course_rating = []

# Extract and store each course rating
for i in soup.find_all("span", class_="urw-din"):
    course_rating.append(i.text)

course_rating
```

## 9. Extract Course Pricing

The pricing information is collected from the `div` elements containing the course price tags.

```python
# Locate the HTML elements containing course prices
soup.find_all("div", class_="courseListingPage_priceTags__PXTc9")
```

```python
# Store the complete pricing text for each course
course_pricing = []

for i in soup.find_all("div", class_="courseListingPage_priceTags__PXTc9"):
    course_pricing.append(i.text)

course_pricing
```

```python
# Split the pricing text to separate discounted and original prices
for i in soup.find_all("div", class_="courseListingPage_priceTags__PXTc9"):
    print(i.text.split("₹"))
```

```python
# Create separate lists for discounted and original prices
disc_price = []
org_price = []

for i in soup.find_all("div", class_="courseListingPage_priceTags__PXTc9"):
    disc_price.append(i.text.split("₹")[1])
    org_price.append(i.text.split("₹")[2])

print("Original Prices:", org_price)
print("Discounted Prices:", disc_price)
```

## 10. Extract Course Thumbnail Image URLs

The image elements for course thumbnails are inspected and their `src` attributes are collected.

```python
# Find all image elements
soup.find_all("img")
```

```python
# Find images specifically used as course thumbnails
soup.find_all("img", alt="course thumbnail")
```

```python
# Inspect the text content of course thumbnail image elements
for i in soup.find_all("img", alt="course thumbnail"):
    print(i.text)
```

```python
# Create a list to store image URLs
img_list = []

for i in soup.find_all("img", alt="course thumbnail"):
    img_list.append(i.get("src"))

img_list
```

## 11. Create a Structured DataFrame

The extracted course name, rating, original price, and discounted price are combined into a Pandas DataFrame.

```python
# Create a Pandas DataFrame from the extracted course information
courses_data = pd.DataFrame({
    "Course Name": course_heading,
    "Course Rating": course_rating,
    "Original Price": org_price,
    "Discount Price": disc_price
})
```

```python
# Display the final structured dataset
courses_data
```

## 12. Export the Data

The final DataFrame is saved as a CSV file so that it can be used for further analysis or processing.

```python
# Save the extracted course data to a CSV file
courses_data.to_csv("GFG_COURSES.csv", index=False)
```

The webpage data was successfully acquired using Requests, parsed using BeautifulSoup, relevant course information was extracted from HTML elements, structured using Pandas, and exported to a CSV file.
