## 2. Word Tokenization

Word tokenization splits a piece of text into individual words (and punctuation marks) called *tokens*.

```python
# Word tokenizer
text = "Ram is a good cook, but he can't make pulses."
tokens = word_tokenize(text)
print(tokens)
```

## 3. Sentence Tokenization

Sentence tokenization splits a paragraph or a document into individual sentences.

```python
# Sentence tokenizer
text = """Ram is a good cook.
He is only 18 years old. His phone number is 87587668"""

sentences = sent_tokenize(text)
sentences
```

## 4. Removing Punctuations

Punctuation marks usually do not add meaning to text analysis, so they are commonly removed during preprocessing.

```python
# Removing punctuations
import string

text = "he is ^%#$ and full of life ?<(*&"
tokens = word_tokenize(text)
print("Before removing punctuations:", tokens)

words = []
for i in tokens:
    if i not in string.punctuation:
        words.append(i)

print("After removing punctuations:", words)
```

## 5. Stop Word Removal

Stop words (like "is", "a", "but", "he") are extremely common words that usually carry little useful information for many NLP tasks, so they are often removed.

```python
# Stop word removal
from nltk.corpus import stopwords
nltk.download('stopwords')

text = "Ram is a good cook, but he can not make pulses."
tokens = word_tokenize(text)

words = []
for i in tokens:
    if i.lower() not in stopwords.words('english'):
        words.append(i)

print("Before removing stopwords:", tokens)
print("After removing stopwords:", words)
```

## 6. Spelling Correction using `pyspellchecker`

Real-world text data (chats, reviews, tweets) often contains spelling mistakes. The `pyspellchecker` library can be used to automatically correct such words.

```python
# Spelling Checking - Method 1: pyspellchecker
!pip install pyspellchecker
from spellchecker import SpellChecker

spell = SpellChecker()

text = "i havv a gud, bd, and ugly situton"
tokens = text.split()

correct = []
for i in tokens:
    correct.append(spell.correction(i))

print("Before correction:", tokens)
print("After correction:", correct)
```

## 7. Spelling Correction using `TextBlob`

`TextBlob` provides another simple way to correct spelling mistakes, this time by correcting the entire string at once instead of word by word.

```python
# Spelling Checking - Method 2: TextBlob
!pip install textblob
from textblob import TextBlob

text = "i havv bot godd fod"
corrected = str(TextBlob(text).correct())

print("Before correction:", text)
print("After correction:", corrected)
```

## 8. Putting It All Together — A Simple Preprocessing Pipeline

The function below combines all the previous steps: spelling correction, tokenization, punctuation removal, and stop word removal.

```python
from spellchecker import SpellChecker
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import string

spell = SpellChecker()

def preprocess_text(text):
    # Step 1: Tokenize
    tokens = word_tokenize(text)

    # Step 2: Correct spelling
    tokens = [spell.correction(word) if spell.correction(word) else word for word in tokens]

    # Step 3: Remove punctuation
    tokens = [word for word in tokens if word not in string.punctuation]

    # Step 4: Remove stop words
    tokens = [word for word in tokens if word.lower() not in stopwords.words('english')]

    return tokens


sample_text = "Ram is a gud cook, but he can't makke pulses!!"
cleaned_tokens = preprocess_text(sample_text)

print("Original text:", sample_text)
print("Cleaned tokens:", cleaned_tokens)
```
